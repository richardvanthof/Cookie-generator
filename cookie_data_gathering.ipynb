```python
import json
import requests

url = "https://api.spoonacular.com/recipes/complexSearch?query=cookies&diet=vegetarian&type=dessert&instructionsRequired=true&number=100&offset=0&includeIngredients=true&fillIngredients=true&addRecipeInstructions=true&addRecipeNutrition=true&apiKey=d43da624d9014c85861eda649252ba83"
resp = requests.get(url)
cookies = resp.json()

with open("cookies_raw.json", "w", encoding="utf-8") as f:
    json.dump(cookies, f, indent=4, ensure_ascii=False)

clist = []
for recipe in cookies.get("results", []):
    rdict = {
        "name": recipe.get("title"),
        "family": "cookie",
        "servings": recipe.get("servings"),
        "ingredients": []
    }
    for ing in recipe.get("extendedIngredients", []):
        metric = ing.get("measures", {}).get("metric", {})
        rdict["ingredients"].append({
            "name": ing.get("name"),
            "amount": metric.get("amount"),
            "unit": metric.get("unitShort"),
            "type": ing.get("aisle"),
            "cons": ing.get("consistency")
        })
    clist.append(rdict)

with open("cookies.json", "w", encoding="utf-8") as f:
    json.dump(clist, f, indent=4, ensure_ascii=False)
```
